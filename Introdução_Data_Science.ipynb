```python
contagem_linguagem= tmdb['original_language'].value_counts().to_frame().reset_index()
contagem_linguagem.columns= ['original_language', 'total']
contagem_linguagem.plot()
```

```python
sns.barplot(data= contagem_linguagem, x="original_language", y="total")
```

```python
#realizando de forma mais rápida
sns.countplot(data= tmdb, x= "original_language")
```

```python
total_por_linguagem= tmdb['original_language'].value_counts()
```

```python
total_de_ingles= total_por_linguagem.loc['en']
total_geral= total_por_linguagem.sum()
total_de_ingles
total_resto= total_geral - total_de_ingles
total_resto

print(total_geral, total_de_ingles, total_resto)
```

```python
dados = {
  "Língua": ["Inglês", "Outros"],
  "Total": [total_de_ingles, total_resto]
}
dados = pd.DataFrame(dados)
dados
```

```python
sns.barplot(data=dados, x= "Língua", y= "Total")
```

```python
dados.plot(kind= "pie", y= "Total", labels= dados['Língua'])
```

```python
total_filmes_lingua_outros= tmdb.query("original_language != 'en'")["original_language"].value_counts()
total_filmes_lingua_outros.head()
```

```python
sns.countplot(data= tmdb.query("original_language !='en'"), x= "original_language")
```

# Refinando uma visualização

```python
sns.color_palette("mako")
```

```python
tmdb.query("original_language != 'en'")["original_language"].value_counts(normalize= True)
```

```python
plt.figure(figsize= (16, 8))

sns.countplot(data= tmdb.query("original_language !='en'"),
              x= "original_language",
              order = total_filmes_lingua_outros.index,
              hue= "original_language",
              hue_order= total_filmes_lingua_outros.index,
              palette= "mako",
              stat= "percent",
              )
plt.title("Distribuição da língua original nos filmes (exceto em inglês)")
plt.show()
```

```python
notas_toy_story= notas.query("FilmeId == 1")["Nota"]
notas_jumanji= notas.query("FilmeId == 2")["Nota"]

media_toy_story= notas_toy_story.mean()
media_jumanji= notas_jumanji.mean()

print(media_toy_story, media_jumanji)
```
